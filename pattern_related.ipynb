```python
rows=int(input("Enter the row's number:"))
for i in range(1, rows+1):
    print("* "*i)
```

```python
rows=int(input("Enter the rows's number:"))
for i in range(rows):
    print(" " * (rows-i-1)+"*"*(2*i+1))
```

```python
rows=int(input("Enter the number of rows:"))
for i in range(rows):
    print(" "*(rows-i-1)+"*"*(2*i+1))

for i in range(rows-2,-1,-1):
    print(" "*(rows-i-1)+"*"*(2*i+1))
```

```python
rows=int(input("Enter the number of rows="))
num=1
for i in range(1,rows+1):
    for j in range(i):
        print(num,end=" ")
        num+=1
    print()
```

```python
rows=int(input("Enter the number:"))
for i in range(rows):
    num=1
    print(" "*(rows-i),end=" ")
    for j in range(i+1):
        print(num,end=" ")
        num=num*(i-j)//(j+1)
    print()
```
