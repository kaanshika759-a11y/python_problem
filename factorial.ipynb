```python
def factorial(n):
    if n==0 or n==1:
        return 1
    return n*factorial(n-1)
n=int(input("Enter the  number:"))
if n<0:
    print("Invalid")
else:
    print("Factorial number=",factorial(n))
      
        
    
```
